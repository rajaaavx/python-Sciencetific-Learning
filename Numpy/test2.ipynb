```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return 15*x**5 + 6*x**4 - 9
def g(x):
    return 45*x**4 + 24*x**3

def newton(x0, max = 50, epsilon = 0.032):
    step = 1
    flag = 1
    condition = True

    while condition:
        x1 = x0 - f(x0)/g(x0)
        x0 = x1
        step += 1

        if step > max:
            flag = 0
            break
        condition = np.abs(f(x1)) < epsilon

    if flag == 1:
        print("converged")
    else:
        print("did not converged")
    
newton(3)
```
