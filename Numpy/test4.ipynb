```python
import numpy as np

Xs = [
    [
        [10, 2, 1],    
        [1, -8, 2],    
        [2, 3, -9]     
    ],
    [ 
        [-12, 2, -3],  
        [3, 9, 1],     
        [-2, 1, 10]    
    ],
    [
        [11, -1, 3],   
        [2, 7, -2],    
        [3, 2, 9]      
    ],
    [  
        [-10, 5, 1],   
        [-4, 11, -3],  
        [-2, 3, -8]    
    ],
    [  
        [8, 1, -2],    
        [-3, 10, 2],   
        [-1, -2, 9]    
    ]
]

Ys = [
    [40, 1150, 18],
    [25, -10, 15],
    [30, -10, 12],
    [80, -700, 140],
    [15, 10, -5]
]
```

```python
def isDom(x):
    x = np.array(x)
    diag = np.diag(np.abs(x))
    off_diag = np.sum(np.abs(x), axis= 1) - diag

    return np.all(diag > off_diag)

def gauss(x, y, max = 8, epsilon = 0.032):
    if not isDom(x):
        print("is not diagonally dominant")
        return
    
    x = np.array(x)
    y = np.array(y)

    diag = np.array(np.diag(x))
    np.fill_diagonal(x, 0)
    old = np.zeros(diag.size)

    for i in range(max):
        new = np.array(old)
        for j, row in enumerate(x):
            new[j] = (y[j] - np.dot(new , row)) / diag[j]
        
        eul = np.sqrt(np.dot(new-old, new-old))

        print(f'iteration-{i}: {new}')

        if(eul < epsilon):
            print("Converged")
            return
        old = new
    
    print("No solution found")

for i in range(len(Xs)):
    gauss(Xs[i], Ys[i])
```
