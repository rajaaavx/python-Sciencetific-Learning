```python
import numpy as np
import matplotlib as plt
```

```python
#Fungtion Diagonaly Dominant

def Diagonaly_dominant(x):
    x = np.array(x)
    diag = np.abs(np.diag(x))

    #axis = 1 -> baris
    #axis = 0 -> kolom
    off_diag = np.sum(np.abs(x), axis=1) - diag 

    return np.all(diag > off_diag) #return must be true or false


```

```python
def gauss_seidel(x, y, max_itr= 62, epsilon= 0.01):
    if not Diagonaly_dominant(x):
        print("Is Not Diagonaly Dominant")
        return
    
    #convert this mfs
    x = np.array(x)
    y = np.array(y)

    diag = np.array(np.diag(x))

    np.fill_diagonal(x, 0)

    #set nilai x1, x2, x3 jd 0
    mtx_old = np.zeros(diag.size)

    for i in range(max_itr):
        mtx_new = np.array(mtx_old)

        for idx, row in enumerate(x):
            mtx_new[idx] = (y[idx] - np.dot(mtx_new, row)) / diag[idx]
        

        print(f'iterasi ke-{i+1}: {mtx_new}')

        d = np.sqrt(np.dot(mtx_new - mtx_old, mtx_new - mtx_old))

        if d < epsilon:
            print("Converged")
            return
        
        mtx_old = mtx_new 

    print("Solution Not Found")   
```

```python
#snipset Persamaan binus
Xs = [
    [10, 4, 5],
    [1, 6, 1],
    [3, 1, 6]
]

Ys = [8, 5, 10]

gauss_seidel(Xs, Ys)
```
