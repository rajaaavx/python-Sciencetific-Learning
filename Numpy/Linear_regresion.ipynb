```python
import numpy as np
import matplotlib.pyplot as plt

x = [1,3,5,7,8,9]
y = [3,5,4,2,1,6]
```

```python

x = np.array(x)
y = np.array(y)

A = np.vstack([x, np.ones(len(x))]).T
y = y[:, np.newaxis]
result = np.dot(np.dot(np.linalg.inv(np.dot(A.T, A)), A.T), y)

m = x*result[0] + result[1]

plt.plot(x, y, 'bo', label= "data")
plt.plot(x, m, 'r', label= "Linear-Regression")
plt.legend()
```
