```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**4 - 13*x**3 + 36

a = -2
b = 2
n = 50
L = (b-a)/n

x = np.linspace(a, b, n)
y = f(x)

left =  L * sum(y[:n-1])
right = L * sum(y[1:])

Xm = (x[:n-1] + x[1:])/2
Ym = f(Xm)
mid = L * sum(Ym)

traps= L/2*(y[0] + 2 * sum(y[1:n-1]) + y[n-1])
print(left)
print(right)
print(mid)
print(traps)
```
