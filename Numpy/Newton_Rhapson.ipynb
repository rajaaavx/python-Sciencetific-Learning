```python
import numpy as np

def f(x):
    return 15*x**5 + 6*x**4 - 9
def g(x):
    return 45*x**4 + 24*x**3 - 9

def newton(x0, max, e):
    step = 1
    flag = 1
    condition = True

    while condition:
        step += 1
        x1 = x0 - f(x0)/g(x0)
        x0 = x1
        condition = np.abs(f(x1)) < e
        if(step > max):
            flag = 0
            break
    
    if flag == 1:
        print("converged")
    else:
        print("did not converged")

newton(3, 25, 0.032)
```
